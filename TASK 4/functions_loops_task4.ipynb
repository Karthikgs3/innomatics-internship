```python
def recharge_validation():
    valid_plans = [199, 299, 399, 599]

    while True:
        amount = int(input("Enter recharge amount: "))

        if amount < 50:
            print("Recharge must be at least ₹50")
            continue

        if amount in valid_plans:
            print("Recharge successful")
            break
        else:
            print("Invalid plan. Try again")


recharge_validation()
```

```python
def check_inventory(products):
    for item, qty in products.items():
        if qty < 15:
            print(f"{item}: Reorder Alert")
        else:
            print(f"{item}: Stock OK")


products = {
    "Apples": 10,
    "Bananas": 25,
    "Oranges": 12
}

check_inventory(products)
```

```python
def student_result(marks):
    total = 0
    for m in marks:
        total += m

    avg = total / len(marks)

    if avg >= 50:
        print("Pass")
    else:
        print("Fail")


student_result([60, 55, 40, 70])
```

```python
def cab_fare():
    while True:
        km = float(input("Enter distance (km): "))
        peak = input("Peak hour? (yes/no): ").lower()

        fare = 50 + (km * 12)

        if peak == "yes":
            fare *= 1.25

        print("Total Fare:", fare)

        retry = input("Calculate again? (yes/no): ").lower()
        if retry != "yes":
            break


cab_fare()
```

```python
def attendance_check(att):
    present = 0

    for i in att:
        if i == "P":
            present += 1

    percentage = (present / len(att)) * 100

    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"


result = attendance_check(["P", "P", "A", "P", "P"])
print(result)
```

```python
def check_password(password):
    if len(password) < 8:
        print("Weak password")
        return

    has_digit = False
    has_special = False

    for ch in password:
        if ch.isdigit():
            has_digit = True
        if ch in "@#$":
            has_special = True

    if has_digit and has_special:
        print("Strong password")
    else:
        print("Weak password")


check_password("Pass@123")
```
