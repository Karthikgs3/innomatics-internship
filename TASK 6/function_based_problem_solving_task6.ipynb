```python
def parking_system(capacity, logs):
    parked = 0
    peak = 0

    for log in logs:
        if log == "IN":
            parked += 1
        elif log == "OUT" and parked > 0:
            parked -= 1

        if parked > peak:
            peak = parked

    print("Currently Parked Vehicles:", parked)

    if parked > capacity:
        print("Parking Status: Over Capacity Alert")
    else:
        print("Parking Status: Available")

    print("Peak Usage:", peak)


logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]
parking_system(50, logs)
```

```python
def delivery_time(distance, traffic, weather):
    eta = distance * 5   # base 5 minutes per km

    if traffic.lower() == "high":
        eta += 15
    elif traffic.lower() == "medium":
        eta += 8

    if weather.lower() == "rainy":
        eta += 10

    print("Estimated Delivery Time:", eta, "minutes")


delivery_time(8, "High", "Rainy")
```

```python
def seat_occupancy(total_seats, booked_list):
    booked = len(booked_list)
    occupancy = (booked / total_seats) * 100

    print(f"Occupancy: {occupancy:.0f}%")

    if occupancy == 100:
        print("Show Status: Housefull")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Seats Available")


seat_occupancy(200, [1]*150)
```

```python
def server_status(cpu_readings):
    total = 0
    for value in cpu_readings:
        total += value

    avg = total / len(cpu_readings)
    print(f"Average CPU Load: {avg:.0f}%")

    if avg < 50:
        print("Server Status: Normal")
    elif 50 <= avg <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")


server_status([45, 60, 70, 85, 90])
```

```python
def resource_monitor(resources):
    overused = []

    for name, hours in resources.items():
        if hours > 8:
            overused.append(name)

    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Energy Alert: No")


resources = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

resource_monitor(resources)
```

```python
def registration_control(capacity, registrations):
    if registrations <= capacity:
        print("Confirmed Registrations:", registrations)
        print("Waitlisted Users: 0")
        print("Registration Status: Open")
    else:
        waitlist = registrations - capacity
        print("Confirmed Registrations:", capacity)
        print("Waitlisted Users:", waitlist)
        print("Registration Status: Closed")


registration_control(100, 105)
```
